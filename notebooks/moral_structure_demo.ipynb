## Part 1: Understanding the Schema

A `MoralBond` captures a normative relationship with these components:

- **bond_type**: obligation, prohibition, permission, claim, etc.
- **agent_role**: who has the duty/right (parent, child, ruler, etc.)
- **patient_role**: who is affected
- **action**: what action is involved (honor, protect, harm, etc.)
- **modal_strength**: how strong (must, should, may)
- **context**: domain (family, economic, religious, etc.)

````python
# Example: Create a moral bond manually
bond = MoralBond(
    bond_type=BondType.OBLIGATION,
    agent_role=RoleType.CHILD,
    patient_role=RoleType.PARENT,
    action=ActionCategory.HONOR,
    modal_strength=ModalStrength.MUST,
    context=ContextType.FAMILY,
    source_tradition="biblical",
    source_language="hebrew",
)

print("Bond as dict:")
print(bond.to_dict())
print()
print("Canonical tuple (for mathematical analysis):")
print(bond.to_canonical_tuple())
````

## Part 3: Mathematical Analysis

Now let's analyze these bonds for patterns.

````python
# Generate full analysis report
report = generate_analysis_report(demo_bonds)
print(report)
````

````python
# Find universal patterns
patterns = find_universal_patterns(demo_bonds, min_cultures=2)

print("\n=== Universal Patterns (2+ cultures) ===")
for p in patterns:
    print(f"\n{p.name}")
    print(f"  Description: {p.description}")
    print(f"  Cultures: {', '.join(sorted(p.cultures_present))}")
    print(f"  Frequency: {p.frequency:.1%}")
````

````python
# Analyze symmetry
symmetry = compute_symmetry_group(demo_bonds)

print("\n=== Symmetry Analysis ===")
print(f"Reciprocal pairs: {len(symmetry['reciprocal_pairs'])}")
if symmetry['reciprocal_pairs']:
    for b1, b2 in symmetry['reciprocal_pairs'][:3]:
        print(f"  {b1.agent_role.value} → {b1.patient_role.value} : {b1.action.value}")
        print(f"  {b2.agent_role.value} → {b2.patient_role.value} : {b2.action.value}")

print(f"\nReflexive bonds: {len(symmetry['reflexive_bonds'])}")
print(f"Inverse pairs (obligation/prohibition): {len(symmetry['inverse_pairs'])}")
````

````python
# Bond algebra statistics
stats = compute_bond_algebra(demo_bonds)

print("\n=== Bond Algebra ===")
print(f"Total bonds: {stats['total_bonds']}")
print(f"Unique canonical patterns: {stats['unique_canonical_tuples']}")

print("\nBond type distribution:")
for bt, count in stats['bond_type_distribution'].most_common():
    print(f"  {bt}: {count}")

print("\nAction distribution:")
for action, count in stats['action_distribution'].most_common(5):
    print(f"  {action}: {count}")
````

## Part 5: Canonical Tuple Comparison

The power of this approach is that canonical tuples allow direct mathematical comparison
across languages without any embedding bias.

````python
# Compare canonical forms
print("=== Canonical Tuple Comparison ===")
print("\nBonds about honoring parents:")
honor_bonds = [b for b in demo_bonds if b.action == ActionCategory.HONOR]

for bond in honor_bonds:
    print(f"\n{bond.source_tradition} ({bond.source_language}):")
    print(f"  Canonical: {bond.to_canonical_tuple()}")
    print(f"  Modal: {bond.modal_strength.value}")
    if bond.condition:
        print(f"  Condition: {bond.condition}")

# Check if they match
tuples = [b.to_canonical_tuple() for b in honor_bonds]
print(f"\nAll same canonical form: {len(set(tuples)) == 1}")
print(f"Unique forms: {len(set(tuples))}")
````

## Why This Works Better Than Embeddings

1. **No language leakage**: Canonical tuples have no implicit language information
2. **Interpretable**: We can see exactly what structure was extracted
3. **Mathematically tractable**: Tuples can be compared, grouped, counted directly
4. **Cross-lingual by design**: The schema is language-independent

The key insight is that **moral structures are concepts**, not linguistic artifacts.
By extracting the concept symbolically, we bypass the language encoding problem entirely.
